```python
import pandas as pd 
import numpy as np
from sklearn.model_selection import train_test_split 
from sklearn.metrics import accuracy_score, classification_report 

from sklearn.neighbors import KNeighborsClassifier
data=pd.read_csv("6iris.csv",header='infer').values 
X=data[:,1:-1]
y=data[:,-1] 
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,stratify=y)
k=int(input("Enter the number of nearest neighbours to be used, i.e. k:"))

model=KNeighborsClassifier(n_neighbors=k, weights='distance')

model.fit(X_train,y_train)

pred=model.predict(X_test)

accuracy=accuracy_score(y_test,pred)
print("Accuracy:", accuracy)
print(classification_report(y_test,pred))
```

```python
import numpy as np
import pandas as pd
from collections import defaultdict
# -----------------------------
# Step 1: Load Iris Dataset
# -----------------------------

url = "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"
columns = ["sepal_length","sepal_width","petal_length","petal_width","class"]
data = pd.read_csv(url, names=columns)
```

```python
data = data.sample(frac=1, random_state=42).reset_index(drop=True)

train_size = int(0.8 * len(data))
train_data = data.iloc[:train_size]
test_data = data.iloc[train_size:]
X_train = train_data.iloc[:, :-1].values
y_train = train_data.iloc[:, -1].values
X_test = test_data.iloc[:, :-1].values
y_test = test_data.iloc[:, -1].values
```

```python
def euclidean_distance(x1, x2):
    return np.sqrt(np.sum((x1 - x2) ** 2))
```

```python
def predict(X_train, y_train, test_point, k=5):
    distances = []
    for i in range(len(X_train)):
        dist = euclidean_distance(test_point, X_train[i])
        distances.append((dist, y_train[i]))
        distances.sort(key=lambda x: x[0])
        neighbors = distances[:k]
        class_weights = defaultdict(float)
        
    for dist, label in neighbors:
        weight = 1 / (dist + 1e-5) # avoid division by zero
        class_weights[label] += weight
    return max(class_weights, key=class_weights.get)
```

```python
predictions = []
for test_point in X_test:
    pred = predict(X_train, y_train, test_point, k=5)
    predictions.append(pred)

correct = 0
for i in range(len(predictions)):
    if predictions[i] == y_test[i]:
        correct += 1
    accuracy = correct / len(y_test)
print("Accuracy:", accuracy)
```

```python
print(classification_report(y_test,predictions))
```

```python
import numpy as np
from collections import defaultdict
def classification_report(y_true, y_pred):
    classes = np.unique(y_true)
    report = {}
    for c in classes:
        tp = 0
        fp = 0
        fn = 0
        tn = 0
        
    for i in range(len(y_true)):
        if y_true[i] == c and y_pred[i] == c:
            tp += 1
        elif y_true[i] != c and y_pred[i] == c:
            fp += 1
        elif y_true[i] == c and y_pred[i] != c:
            fn += 1
        else:
            tn += 1
        precision = tp / (tp + fp) if (tp + fp) != 0 else 0
        recall = tp / (tp + fn) if (tp + fn) != 0 else 0
        if precision + recall == 0:
            f1 = 0
        else:
            f1 = 2 * precision * recall / (precision + recall)
            support = sum(y_true == c)
            report[c] = [precision, recall, f1, support]
    print("Class\t\tPrecision\tRecall\tF1-score\tSupport")
    for c, values in report.items():
        print(f"{c}\t{values[0]:.2f}\t\t{values[1]:.2f}\t{values[2]:.2f}\t\t{values[3]:.2f}\t\t")
```

```python
classification_report(y_test, np.array(predictions))
```
